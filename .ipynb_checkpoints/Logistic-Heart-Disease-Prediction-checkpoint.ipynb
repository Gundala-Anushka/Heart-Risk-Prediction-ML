### **Importing the Dependencies**

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

### **Data collecting and processing**

```python
# Loading the csv data to a Pandas dataframe
heart_disease_data = pd.read_csv('heart_risk_dataset.csv')
```

```python
# Printing first 5 rows of the dataset
heart_disease_data.head()
```

```python
# Printing last 5 rows of the dataset
heart_disease_data.tail()
```

```python
# Printing the number of rows and columns in the dataset
heart_disease_data.shape
```

```python
# Getting some info about the dataset
heart_disease_data.info()
```

```python
# Checking for missing values
heart_disease_data.isnull().sum()
```

```python
# Statistical measures about the dataset
heart_disease_data.describe()
```

```python
print(heart_disease_data.columns)
```

```python
# Checking the distribution of target variable
heart_disease_data['heart_risk'].value_counts()
```

1 --> Defective Heart

0 --> Healthy Heart

### **Splitting the Features and Targets**

```python
X = heart_disease_data.drop(columns='heart_risk')
Y = heart_disease_data['heart_risk']
```

```python
print(X)
```

```python
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

### **Model training**

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

model = make_pipeline(StandardScaler(), LogisticRegression())
```

```python
# Training the Logistic Regression model with training data
model.fit(X_train, Y_train)
```

### **Model evaluation**

```python
# Calculating accuracy score on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy on training data: ', training_data_accuracy)
```

```python
# Calculating accuracy score on testing data
X_test_prediction = model.predict(X_test)
testing_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy on testing data: ', testing_data_accuracy)
```

### **Building predictive system**

```python
# Automatically grab the 16 feature names from your X dataframe
list_features = list(X.columns)

# Option A: Test using the first patient from your dataset (16 features)
input_data = tuple(X.iloc[0].values)

# Create a DataFrame using the input data and feature names
input_df = pd.DataFrame([input_data], columns=list_features)

prediction = model.predict(input_df)
print("Prediction:", prediction)

if prediction[0] == 0:
    print('The Person does not have a Heart Disease (Low Risk)')
else:
    print('The Person has Heart Disease (High Risk)')
```

```python
import joblib
joblib.dump(model, 'heart_risk_model.pkl')
print("Saved to VS Code folder!")
```
